```python
from langgraph.graph import StateGraph, START, END
from typing import TypedDict
```

```python
class BatsmanScore(TypedDict):
    runs: int
    balls: int
    fours: int
    sixes: int
    sr: float
    bpb: float
    boundary_percentage: float
    summary: str
```

```python
def calculate_sr(state: BatsmanScore):
    sr = state['sr'] = (state['runs'] / state['balls']) * 100
    return {'sr':sr}
```

```python
def calculate_bpb(state: BatsmanScore):
    total_boundaries = state['fours'] + state['sixes']
    bpb= state['balls'] / total_boundaries if total_boundaries > 0 else 0.0
    return {'bpb':bpb}
```

```python
def calculate_boundary_percent(state: BatsmanScore):
    boundary_runs = (state['fours'] * 4) + (state['sixes'] * 6)
    percent = (boundary_runs / state['runs']) * 100 if state['runs'] > 0 else 0.0
    return {'boundary_percentage':percent}
```

```python
def summary(state: BatsmanScore):
   summary= f"""
Strike Rate - {state['sr']:.2f}
Balls per Boundary - {state['bpb']:.2f}
Boundary Percentage - {state['boundary_percentage']:.2f}%
"""
   return {'summary':summary}
```

```python
graph = StateGraph(BatsmanScore)

graph.add_node("calculate_sr", calculate_sr)
graph.add_node("calculate_bpb", calculate_bpb)
graph.add_node("calculate_boundary_percent", calculate_boundary_percent)
graph.add_node("summary", summary)
```

```python
graph.add_edge(START, "calculate_sr")
graph.add_edge(START, "calculate_bpb")
graph.add_edge(START, "calculate_boundary_percent")

graph.add_edge("calculate_sr", "summary")
graph.add_edge("calculate_bpb", "summary")
graph.add_edge("calculate_boundary_percent", "summary")

graph.add_edge("summary", END)
workflow = graph.compile()
```

```python
initial_state = {
    "runs": 75,
    "balls": 50,
    "fours": 7,
    "sixes": 3,
    "sr": 0.0,
    "bpb": 0.0,
    "boundary_percentage": 0.0,
    "summary": ""
}

result = workflow.invoke(initial_state)
print(result)
print(result["summary"])
```
