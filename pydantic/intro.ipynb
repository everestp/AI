```python
print(1)
from pydantic import BaseModel

class Person1(BaseModel):
    name:str
    age:int
    city:str

person =Person1(name="Krish",age=35 ,city="Banglore")
print(person)
person1=Person1(name="Krish",age=35 , city=12)
print(person1)
```

```python
from dataclasses import dataclass

@dataclass
class Person2():
    name:str
    age:int
    city:str

person=Person2(name="Krish",age=35 , city="Bagalore")
print(person)
```

```python
from typing import List


class ClassRoom(BaseModel):
    room_number:str
    students: List[str]
    capacity: int
    classroom = ClassRoom(
    room_number="A101",
    students=("Everest","Rest"),
    capacity=30
)
print(classroom)
```
